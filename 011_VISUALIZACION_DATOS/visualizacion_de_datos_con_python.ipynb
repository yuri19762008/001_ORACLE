# Visualización de datos con Python: Matplotlib, Seaborn y Plotly

```python
import pandas as pd

df = pd.read_csv('inmigrantes_canada.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.set_index('Pais', inplace=True)
df
```

```python
anos = list(map(str, range(1980, 2014)))
anos
```

```python

# pandas series 
colombia = df.loc['Colombia', anos]
colombia
```

```python
col_dict = {'Año': colombia.index.tolist(),
           'Inmigrantes': colombia.values.tolist()}
col_dict
```

```python
datos_col = pd.DataFrame(col_dict)
datos_col
```

# GRAFICANDO CON MATPLOTLIB

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)
plt.title('Inmigrantes colombianos en Canadá (1980-2013)')
plt.xlabel('Año')
plt.ylabel('Número de inmigrantes')
plt.show() #se deja siempre de ultimo para que se muestre la gráfica
```

```python
colores = ['royalblue', 'orange', 'forestgreen','orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy','teal', 'tomato']


fig, ax = plt.subplots(figsize=(12, 5))

ax.barh(sudamerica.index, sudamerica['Total'], color=colores, alpha=0.7)
ax.set_title('Inmigración de países sudamericanos a Canadá\n (1980-2013)', fontsize=16, fontweight='bold', color='black',loc='left')

ax.set_ylabel('', fontsize=12, fontweight='bold', color='red')
ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
ax.xaxis.set_tick_params(labelsize=10)
ax.yaxis.set_tick_params(labelsize=8)

plt.show()
```

```python
sudamerica_sorted = sudamerica.sort_values(by='Total', ascending=True)

colores = ['royalblue', 'orange', 'forestgreen','orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy','teal', 'tomato']


fig, ax = plt.subplots(figsize=(12, 5))

ax.barh(sudamerica_sorted.index, sudamerica_sorted['Total'], color=colores, alpha=0.7)
ax.set_title('Inmigración de países sudamericanos a Canadá, (1980-2013)', fontsize=16, fontweight='bold', color='black',loc='left')

ax.set_ylabel('', fontsize=12, fontweight='bold', color='red')
ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
ax.xaxis.set_tick_params(labelsize=10)
ax.yaxis.set_tick_params(labelsize=8)

plt.show()
```

```python
colores =[]
for pais in sudamerica_sorted.index:
    if pais == 'Colombia':
        colores.append('forestgreen')
    else:
        colores.append('silver')
        
        
fig, ax = plt.subplots(figsize=(12, 5))

ax.barh(sudamerica_sorted.index, sudamerica_sorted['Total'], color=colores, alpha=0.7)
ax.set_title('Inmigración de Colombianos a Canadá, (1980-2013)', fontsize=16, fontweight='bold', color='black',loc='left')

ax.set_ylabel('', fontsize=12, fontweight='bold', color='red')
ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
ax.yaxis.set_tick_params(labelsize=10)
ax.xaxis.set_tick_params(labelsize=8)

plt.show()
```

```python
#cambia color a colombia en el gráfico de barras
colores =[]
for pais in sudamerica_sorted.index:
    if pais == 'Colombia':
        colores.append('forestgreen')
    else:
        colores.append('silver')
        
        
fig, ax = plt.subplots(figsize=(12, 5))

ax.set_frame_on(False)  # Elimina el marco del gráfico
ax.xaxis.set_visible(False)  # Elimina los ejes
ax.tick_params(axis='y', which='both', left=False, right=False)  # Elimina las marcas de los ejes y etiquetas

ax.barh(sudamerica_sorted.index, sudamerica_sorted['Total'], color=colores, alpha=0.7)
ax.set_title('Inmigración de Colombianos a Canadá, (1980-2013)', fontsize=16, fontweight='bold', color='black',loc='left')

ax.set_ylabel('', fontsize=12, fontweight='bold', color='red')
ax.yaxis.set_tick_params(labelsize=10)
#ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
#ax.xaxis.set_tick_params(labelsize=8)

#coloca el valor de cada barra al lado
for i,j in enumerate(sudamerica_sorted['Total']):
    ax.text(j + 100, i, str(j), va='center',ha='left', fontsize=9, color='black')
    
# Guarda la figura en formato PNG
fig.savefig('inmigrantes_sudamericanos_canada.png', format='png', dpi=300, bbox_inches='tight', transparent=False)

plt.show()
```

# GRAFICANDO CON SEABORN

```python
import seaborn as sns
```

```python
sns.set_theme()
```

```python
top_10 = df.sort_values(by='Total', ascending=False).head(10)
top_10
```

```python
sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h')

plt.show()
```

```python
ax = sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h')
ax.set(title='Top 10 países con más inmigrantes a Canadá (1980-2013)', xlabel='Número de inmigrantes', ylabel='Países')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))
ax = sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h')
ax.set_title('Top 10 países con más inmigrantes a Canadá (1980-2013)', loc='left', fontsize=16)
ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
ax.set_ylabel(' ', fontsize=12, fontweight='bold', color='red')
plt.show()
```

```python
def generer_grafico(palette):
    fig, ax = plt.subplots(figsize=(12, 5))
    
    ax = sns.barplot(data=top_10, x='Total', y=top_10.index, hue=top_10.index, orient='h',palette=palette)
    
    ax.set_title('Top 10 países con más inmigrantes a Canadá (1980-2013)', loc='left', fontsize=16)
    ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
    ax.set_ylabel(' ', fontsize=12, fontweight='bold', color='red')
    
    for i,j in enumerate(top_10['Total']):
        ax.text(j + 20, i, str(j), va='center', ha='left', fontsize=9, color='black')  
    plt.show()
```

```python
def generer_grafico(palette):
    fig, ax = plt.subplots(figsize=(12, 5))
    ax.set(xticklabels=[])
    
    ax = sns.barplot(data=top_10, x='Total', y=top_10.index, hue=top_10.index, orient='h',palette=palette)
    
    ax.set_title('Top 10 países con más inmigrantes a Canadá (1980-2013)', loc='left', fontsize=16)
    ax.set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
    ax.set_ylabel(' ', fontsize=12, fontweight='bold', color='red')
    sns.despine()
    
    for i,j in enumerate(top_10['Total']):
        ax.text(j + 20, i, str(j), va='center', ha='left', fontsize=9, color='black')  
    plt.show()
```

```python
sns.set_theme()
sns.set_palette('Dark2')

fig, ax = plt.subplots(figsize=(10, 5))
ax = sns.lineplot(df.loc['Brasil', anos], label='Brasil', lw=3)
ax = sns.lineplot(df.loc['Argentina', anos], label='Argentina', lw=3)
ax = sns.lineplot(df.loc['Perú', anos], label='Perú', lw=3)
ax = sns.lineplot(df.loc['Colombia', anos], label='Colombia', lw=3)

ax.set_title('Inmigración de los principales países de América del Sur\nal Canadá de 1980 a 2013', loc='left', fontsize=20)
ax.set_xlabel('Año', fontsize=14)
ax.set_ylabel('Número de Inmigrantes', fontsize=14)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.legend(title='Países', loc='upper right', bbox_to_anchor=(1.18, 1.02))

plt.show()
```
