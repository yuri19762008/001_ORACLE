# Visualización de datos con Python: Matplotlib, Seaborn y Plotly

```python
import pandas as pd

df = pd.read_csv('inmigrantes_canada.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.set_index('Pais', inplace=True)
df
```

```python
anos = list(map(str, range(1980, 2014)))
anos
```

```python

# pandas series 
colombia = df.loc['Colombia', anos]
colombia
```

```python
col_dict = {'Año': colombia.index.tolist(),
           'Inmigrantes': colombia.values.tolist()}
col_dict
```

```python
datos_col = pd.DataFrame(col_dict)
datos_col
```

### GRAFICANDO CON MATPLOTLIB

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)
plt.title('Inmigrantes colombianos en Canadá (1980-2013)')
plt.xlabel('Año')
plt.ylabel('Número de inmigrantes')
plt.show() #se deja siempre de ultimo para que se muestre la gráfica
```

## ejercicio 
- En clase solo extrajimos los datos de Colombia, pero ahora podemos ejecutar el siguiente comando para extraer los datos de Brasil y Argentina:

```python
df_comparacion = df.loc[['Brasil', 'Argentina'], anos]
```

```python
df_comparacion = df_comparacion.T
```

```python
df_comparacion.head()
```

```python
plt.figure(figsize=(14, 4))
plt.plot(df_comparacion['Brasil'], label='Brasil')
plt.plot(df_comparacion['Argentina'], label='Argentina')
plt.title('Inmigración de Brasil y Argentina a Canadá')
plt.xlabel('Año', fontsize=14, fontweight='bold', color='blue', style='italic')
plt.ylabel('Número de Inmigrantes', fontsize=14, fontweight='bold', color='blue', style='italic')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)

plt.grid()
plt.legend()
plt.show()
```

## CONTINUADO CON LA CREACION DE GRAFICOS

```python
plt.figure(figsize=(10, 4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)
plt.title('Inmigrantes colombianos en Canadá (1980-2013)')
plt.xlabel('Año')
plt.ylabel('Número de inmigrantes')
plt.show() #se deja siempre de ultimo para que se muestre la gráfica
```

```python
fig , ax = plt.subplots(figsize=(10, 4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'], label='Colombia', color='green')
ax.xaxis.set_major_locator(plt.MaxNLocator(5))
ax.set_title('Inmigrantes colombianos en Canadá\n (1980-2013)')
ax.set_xlabel('Años', fontsize=12, fontweight='bold', color='red')
ax.set_ylabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
plt.show()
```

### CONTNUANDO CON GRAFICAR

```python
fig , ax = plt.subplots(figsize=(8, 4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'], label='Colombia', color='green')
ax.xaxis.set_major_locator(plt.MaxNLocator(5))
ax.set_title('Inmigrantes colombianos en Canadá\n (1980-2013)')
ax.set_xlabel('Años', fontsize=12, fontweight='bold', color='red')
ax.set_ylabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
plt.show()
```

```python
fig , axs = plt.subplots(1, 2, figsize=(14, 5))

axs[0].plot(datos_col['Año'], datos_col['Inmigrantes'], label='Colombia', color='green')
axs[0].set_title('Inmigrantes colombianos en Canadá\n (1980-2013)')
axs[0].set_xlabel('Años', fontsize=12, fontweight='bold', color='red')
axs[0].set_ylabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

axs[1].boxplot(datos_col['Inmigrantes'], vert=True)
axs[1].set_title('Boxplot de inmigrantes colombianos en\n Canadá')
axs[1].set_xlabel('Número de inmigrantes', fontsize=12, fontweight='bold', color='blue')
axs[1].set_yticklabels(['Colombia'])
axs[1].grid()

plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 6))
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle('Inmigración a Canadá de países latinoamericanos (1980-2013)', fontsize=16, fontweight='bold', color='black')

axs[0,0].plot(df.loc['Colombia',anos])
axs[0,0].set_title('Colombia')

axs[0,1].plot(df.loc['Brasil',anos])
axs[0,1].set_title('Brasil')

axs[1,0].plot(df.loc['Argentina',anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Perú',anos])
axs[1,1].set_title('Perú')

ymin = 0
ymax = 7000

for ax in axs.ravel():
    ax.set_ylim(ymin, ymax)


for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MaxNLocator(5))
    ax.set_xlabel('Año')
    ax.set_ylabel('Número de inmigrantes')
    ax.set_xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'])
    ax.set_xticklabels(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2013'], rotation=45)
    ax.grid()

plt.show()
```
