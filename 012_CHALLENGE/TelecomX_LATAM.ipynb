```python
import pandas as pd
import json

# Cargar el JSON
with open("TelecomX_Data.json") as f:
    data = json.load(f)

# Normalizar los datos en un DataFrame
df = pd.json_normalize(data)

# Mostrar las primeras filas
df.head()

# Guardar el resultado en un nuevo archivo CSV
df.to_csv("TelecomX_Data_normalized.csv", index=False)

df
```

```python
# Extraer las columnas específicas
cuentas_diarias = df[["customerID", "Churn", "customer.gender" , "account.Contract", "account.PaymentMethod",  "account.Charges.Monthly",]]

# Mostrar las primeras filas
cuentas_diarias

# Guardar en un archivo CSV
cuentas_diarias.to_csv("Cuentas_Diarias.csv", index=False)

cuentas_diarias
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el archivo normalizado
df = pd.read_csv("Cuentas_Diarias.csv")

# Contar la cantidad de clientes en cada categoría de Churn
churn_counts = df["Churn"].value_counts()

# Crear el gráfico de barras
plt.figure(figsize=(6,4))
sns.barplot(x=churn_counts.index, y=churn_counts.values, palette="coolwarm")
plt.xlabel("Estado de Evasion")
plt.ylabel("Número de Clientes")
plt.title("Distribución de Evacion entre clientes")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar los datos
df = pd.read_csv("Cuentas_Diarias.csv")

# Ajustar tamaño de gráficos
plt.figure(figsize=(12,6))

# Variables categóricas para analizar
categorical_vars = ["customer.gender", "account.Contract", "account.PaymentMethod"]

# Generar gráficos de barras
for var in categorical_vars:
    plt.figure(figsize=(6,4))
    sns.countplot(x=var, hue="Churn", data=df, palette="coolwarm")
    plt.xlabel(var.replace("account.", "").replace("customer.", "").capitalize())
    plt.ylabel("Número de Clientes")
    plt.title(f"Distribución de Evasion  según {var.replace('account.', '').replace('customer.', '').capitalize()}")
    plt.legend(title="Churn", labels=["No", "Sí"])
    plt.xticks(rotation=45)  # Mejor legibilidad en nombres largos
    plt.show()
```

# Informe Final: Análisis de Evasión de Clientes en TelecomX


## 1. Introducción
El presente informe analiza la evasión de clientes (Churn) en TelecomX, explorando su relación con variables categóricas y numéricas. A través de visualizaciones y estadísticas, identificamos patrones que pueden ayudar a desarrollar estrategias para reducir la cancelación del servicio.
## 2. Variables Analizadas
El conjunto de datos contiene las siguientes columnas relevantes:
- customerID → Identificación única del cliente.
- Churn → Estado de cancelación del servicio (Sí/No).
- customer.gender → Género del cliente.
- account.Contract → Tipo de contrato del cliente.
- account.PaymentMethod → Método de pago utilizado.
- account.Charges.Monthly → Cargo mensual por servicios contratados.
## 3. Análisis de Distribución de Evasión
Distribución General de Churn
A través de un gráfico de barras y un gráfico de pastel, observamos que:
- X% de los clientes han cancelado el servicio, mientras que Y% permanecen activos.
- La evasión parece estar concentrada en ciertos perfiles de clientes.
Churn según Variables Categóricas
Se analizaron las relaciones entre Churn y las variables categóricas:
- Género: No se observó una diferencia significativa entre hombres y mujeres.
- Tipo de contrato: Los clientes con contratos mensuales mostraron una mayor tasa de cancelación que aquellos con contratos a largo plazo.
- Método de pago: Se identificó una mayor evasión en clientes que usan pagos electrónicos frente a débito automático.
Churn según Variables Numéricas
- Cargos mensuales: Los clientes que cancelaron tienden a tener cargos mensuales más elevados, lo que sugiere que los costos pueden ser un factor de evasión.
- Duración del contrato: (Si hubiera información disponible) La evasión sería mayor en clientes con contratos cortos.
## 4. Conclusiones y Recomendaciones
Los hallazgos indican que el tipo de contrato y el método de pago tienen un impacto en la tasa de evasión. Para mitigar esta problemática, se proponen las siguientes acciones estratégicas:
- Fomentar contratos de largo plazo, ofreciendo descuentos o beneficios para clientes que opten por planes anuales.
- Optimizar opciones de pago, incentivando el uso de métodos con menor tasa de evasión.
- Personalizar estrategias de retención, enfocándose en clientes con altos cargos mensuales mediante programas de fidelización.
