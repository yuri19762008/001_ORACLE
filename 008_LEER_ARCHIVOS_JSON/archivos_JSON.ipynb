### **ejercicio de url**

```python
import requests
import json
```

```python
datos_frutas = requests.get('https://fruityvice.com/api/fruit/all')
```

```python
resultado = json.loads(datos_frutas.text)
```

```python
pd.DataFrame(resultado)
```

guardamos en un data frame resultado

```python

df_resultado = pd.DataFrame(resultado)
df_resultado
```

sacamos el json de la columna nutritions usando la funcion Normalize

```python
df_resultado_norm = pd.json_normalize(df_resultado['nutritions'])

df_resultado_norm 
```
