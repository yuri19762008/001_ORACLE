Regresión lineal

```python
n = np.size(Moscu)
y = Moscu
x = fechas
a = (n * np.sum(x*y) - np.sum(x) * np.sum(y)) / (n * np.sum(x**2) - np.sum(x)**2)
b = np.mean(y) - a * np.mean(x)
print(f'la pendiente es {a} \nla intersección es {b}')
```
