#1. Análisis de facturación



```python
import pandas as pd

# URLs de los archivos CSV
url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

# Cargar los datos de cada tienda en una lista
tiendas = [pd.read_csv(u) for u in [url1, url2, url3, url4]]

# Análisis de facturación para cada tienda
resultados = []
for i, df in enumerate(tiendas, 1):
    ingresos_totales = df['Precio'].sum()
    cantidad_ventas = len(df)
    precio_promedio = df['Precio'].mean()
    max_precio = df['Precio'].max()
    min_precio = df['Precio'].min()
    resultados.append({
        'Tienda': f'Tienda {i}',
        'Ingresos Totales': ingresos_totales,
        'Cantidad de Ventas': cantidad_ventas,
        'Precio Promedio': precio_promedio,
        'Precio Máximo': max_precio,
        'Precio Mínimo': min_precio
    })

# Mostrar resultados en un DataFrame
resultados_df = pd.DataFrame(resultados)
print(resultados_df)
```

# 2. Ventas por categoría

```python
import pandas as pd

# URLs de los archivos CSV
tienda_urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"
]
nombres_tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

# Cargar y unir todos los datos, agregando columna de tienda
dfs = []
for url, nombre in zip(tienda_urls, nombres_tiendas):
    df = pd.read_csv(url)
    df['Tienda'] = nombre
    dfs.append(df)
df_total = pd.concat(dfs, ignore_index=True)

# Agrupar por categoría y sumar ventas (Precio)
ventas_por_categoria = df_total.groupby('Categoría del Producto')['Precio'].sum().sort_values(ascending=False)

print("Ventas totales por categoría (todas las tiendas):")
print(ventas_por_categoria)

# Mostrar ventas por categoría y por tienda
tabla = df_total.groupby(['Categoría del Producto', 'Tienda'])['Precio'].sum().unstack(fill_value=0)
print("\nVentas por categoría y por tienda:")
print(tabla)
```

# 3. Calificación promedio de la tienda


```python
import pandas as pd

# URLs de los archivos CSV
tiendas_urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"
]
nombres_tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

# Calcular calificación promedio de cada tienda
calificaciones = {}
for url, nombre in zip(tiendas_urls, nombres_tiendas):
    df = pd.read_csv(url)
    calificaciones[nombre] = df['Calificación'].mean()

# Mostrar resultados en un DataFrame
calificaciones_df = pd.DataFrame(list(calificaciones.items()), columns=["Tienda", "Calificación Promedio"])
print(calificaciones_df)
```

# 4. Productos más y menos vendidos

```python
import pandas as pd

# URLs de los archivos CSV
tiendas_urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"
]
nombres_tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

# Analizar productos más y menos vendidos por tienda
resultados = []
for url, nombre in zip(tiendas_urls, nombres_tiendas):
    df = pd.read_csv(url)
    conteo = df['Producto'].value_counts()
    mas_vendido = conteo.idxmax()
    menos_vendido = conteo.idxmin()
    resultados.append({
        'Tienda': nombre,
        'Producto Más Vendido': mas_vendido,
        'Cantidad Más Vendido': conteo.max(),
        'Producto Menos Vendido': menos_vendido,
        'Cantidad Menos Vendido': conteo.min()
    })

# Mostrar resultados en un DataFrame
productos_df = pd.DataFrame(resultados)
print(productos_df)
```
