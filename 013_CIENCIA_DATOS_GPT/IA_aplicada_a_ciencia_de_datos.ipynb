````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Supongamos que ya tienes tu DataFrame df
# Por ejemplo:
# df = pd.DataFrame({'metodo_pago': ['Efectivo', 'Tarjeta', 'Transferencia', 'Efectivo', 'Tarjeta', 'Efectivo']})

# Paso 1: Crear la tabla resumen
metodos_de_pago = df['metodo_pago'].value_counts().reset_index()
metodos_de_pago.columns = ['metodo_pago', 'cantidad']

# Mostrar la tabla
print(metodos_de_pago)
````

### **2da Pregunta:** ¿Cuál fue la facturación por categoria em 2023?

**Observación:** Para la facturación, consideraremos aquí el producto entre el precio unitario y la cantidad, sumados al costo de envío de cada producto.


> #### <strong><font color="7b71ff" size=6>Prompt:</font></strong>

Tenemos un DataFrame llamado "df" y queremos construir un gráfico de barras horizontales que muestre la facturación por categoría de Zoop.

En "df", tenemos las siguientes columnas:
* categoria: categorías de los productos
* precio_unitario: precio de una unidad del producto
* cantidad: cantidad vendida del producto
* envio: costo de envío total de la compra.

Deseamos crear una visualización que muestre la facturación de cada categoría, donde la facturación se calcula multiplicando el precio unitario por la cantidad y sumando el costo de envío.

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Crear la columna de facturación
df['facturacion'] = (df['precio_unitario'] * df['cantidad']) + df['envio']

# Paso 2: Agrupar por categoría
facturacion_categoria = df.groupby('categoria')['facturacion'].sum().reset_index()

# Paso 3: Ordenar de mayor a menor (opcional, para mejor visualización)
facturacion_categoria = facturacion_categoria.sort_values(by='facturacion', ascending=True)

# Paso 4: Visualización con gráfico de barras horizontales
plt.figure(figsize=(10, 6))
sns.barplot(data=facturacion_categoria, x='facturacion', y='categoria', palette='viridis')
plt.title('Facturación por Categoría de Producto - Zoop')
plt.xlabel('Facturación Total')
plt.ylabel('Categoría')
plt.tight_layout()
plt.show()
````

<font color="7b71ff" size=4> Comentarios: </font>

---

Aquí especificamos el tipo de gráfico, guiando a ChatGPT en la creación de nuestro visual. Podemos notar que en un gráfico de barras el enfoque está más centrado en las categorías, además de ser una forma más fluida de leer textos que podrían ser demasiado extensos para un gráfico de columnas.

````python
import matplotlib.pyplot as plt

# Agrupar y contar métodos de pago
metodos_pago = df['metodo_pago'].value_counts()

# Calcular el porcentaje de los dos primeros métodos
porcentaje = (metodos_pago.iloc[0] + metodos_pago.iloc[1]) / metodos_pago.sum() * 100

# Colores institucionales de Zoop
ROJO_1, ROJO_2, ROJO_3 = "#e23155", "#cc2c4e", "#b32742"
AZUL_1, AZUL_2, AZUL_3 = "#203f75", "#1c3867", "#19325b"
BLANCO, GRIS_1, GRIS_2, GRIS_3, GRIS_4, GRIS_5 = "#ffffff", "#ebebeb", "#d9d9d9", "#cccccc", "#555655", "#231f20"
AQUA_1, AQUA_2, AQUA_3 = "#addcd4", "#9fccc5", "#96bfb9"

# Crear figura
fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor(GRIS_1)
ax.set_facecolor(GRIS_1)

# Gráfico de barras
barras = ax.bar(metodos_pago.index, metodos_pago.values, color=ROJO_1)

# Título
ax.set_title("Métodos de pago más utilizados en 2023", fontsize=18, color=GRIS_5)

# Eliminar etiquetas y eje Y
ax.set_ylabel("")
ax.set_yticks([])

# Etiquetas eje X
ax.set_xticklabels(metodos_pago.index, fontsize=12, color=AZUL_1)
ax.set_xlabel("", fontsize=12, color=AZUL_1)

# Quitar bordes
for spine in ['top', 'right', 'left']:
    ax.spines[spine].set_visible(False)

# Añadir los valores sobre cada barra
for barra in barras:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width() / 2,
            altura + 1,
            f"{int(altura)}",
            ha='center', fontsize=12, color=AZUL_1)

# Añadir texto explicativo dentro del gráfico
texto = (
    f"$\\bf{{{porcentaje:.2f}}}$% de los clientes utilizan Tarjeta de Crédito o Transferencia \n"
    "para el pago. Sería interesante considerar asociaciones con \n"
    "bancos para la creación de un Zoop Pay con el fin de fidelizar \n"
    "a nuestra clientela."
)

ax.text(1.5, 2500, texto, fontsize=10, color=AZUL_3)

plt.tight_layout()
plt.show()
````

````python
# Supongamos que ya se ha calculado la columna 'facturacion'

# 1. Crear una nueva columna 'mes' que extrae el mes de la columna 'fecha'
df['mes'] = df['fecha'].dt.month_name()

# 2. Agrupar los datos por mes y calcular la facturación total mensual
ventas_mensuales = df.groupby(df['fecha'].dt.to_period('M'))['facturacion'].sum().reset_index()

# 3. Convertir el periodo de fecha a tipo datetime para mantener formato de fecha
ventas_mensuales['fecha'] = ventas_mensuales['fecha'].dt.to_timestamp()

# 4. Crear una columna de nombre del mes en español usando el diccionario proporcionado
meses = {
    'January': 'Ene',
    'February': 'Feb',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'May',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Sep',
    'October': 'Oct',
    'November': 'Nov',
    'December': 'Dic'
}

ventas_mensuales['mes'] = ventas_mensuales['fecha'].dt.month_name().map(meses)

# 5. Crear el gráfico de líneas de ventas mensuales
plt.figure(figsize=(10, 6))
sns.lineplot(x='mes', y='facturacion', data=ventas_mensuales, marker='o', sort=False)

# 6. Añadir etiquetas y título
plt.title('Ventas Totales Mensuales de Zoop', fontsize=16)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Facturación Total', fontsize=12)

# 7. Mostrar el gráfico
plt.show()
````

en color AZUL_1. Añade la caja de anotación con el texto en la posición xytext=(2, 1800) y coloca una flecha, en color AZUL_1, saliendo de la caja hasta la posición xy=(8.7, 2250).

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


def distribucion_calificaciones():
    # Colores identidad visual Zoop
    GRIS_1 = '#E1E1E1'
    GRIS_3 = '#A3A3A3'
    GRIS_5 = '#444444'
    AZUL_1 = '#175676'
    ROJO_1 = '#D72638'
    BLANCO = '#FFFFFF'

    # Estilo general y figura
    sns.set(style="whitegrid")
    fig, ax = plt.subplots(figsize=(10, 6), facecolor=GRIS_1)
    ax.set_facecolor(GRIS_1)
    ax.set_axisbelow(True)

    # Histograma con 11 bins de 0 a 10
    sns.histplot(
        data=df,
        x='nota',
        bins=11,
        kde=False,
        color=ROJO_1,
        ax=ax,
        edgecolor=GRIS_1
    )

    # Título
    ax.set_title('Distribución de las evaluaciones de las compras en 2023', fontsize=18, color=GRIS_5)

    # Ejes y ticks
    ax.set_xlabel('Nota otorgada (0 a 10)', fontsize=12, color=AZUL_1)
    ax.set_ylabel('Cantidad de Evaluaciones', fontsize=12, color=AZUL_1)
    ax.set_xticks(range(0, 11))
    ax.tick_params(axis='x', labelsize=12, colors=AZUL_1)
    ax.tick_params(axis='y', labelsize=12, colors=AZUL_1)

    # Eliminar bordes del gráfico
    for spine in ['top', 'bottom', 'left', 'right']:
        ax.spines[spine].set_visible(False)

    # Anotación
    promedio = df['nota'].mean().round(2)
    moda = df['nota'].value_counts().idxmax()
    frecuencia_moda = df['nota'].value_counts().max()

    texto_anotacion = (
        f"Promedio de calificaciones = {promedio}\n"
        f"Calificación más común = {moda}\n"
        f"Frecuencia más común = {frecuencia_moda}"
    )

    ax.annotate(
        texto_anotacion,
        xy=(8.7, 2250),             # Punto que señala la flecha
        xytext=(2, 1800),          # Posición de la caja de texto
        fontsize=12,
        color=AZUL_1,
        arrowprops=dict(facecolor=AZUL_1, arrowstyle='->'),
        bbox=dict(boxstyle="round,pad=0.4", fc=BLANCO, ec=GRIS_3, lw=1.5)
    )

    plt.tight_layout()
    return plt
````

````python
distribucion_calificaciones().show()
````

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def box_plot_edad_sexo():
    # Definición de colores identidad Zoop
    GRIS_1 = '#E1E1E1'
    GRIS_3 = '#A3A3A3'
    GRIS_5 = '#444444'
    AZUL_1 = '#175676'
    AQUA_1 = '#5FB7D4'
    BLANCO = '#FFFFFF'

    # Estilo general
    sns.set(style="whitegrid")

    # Crear figura y ejes
    fig, ax = plt.subplots(figsize=(10, 6), facecolor=GRIS_1)
    ax.set_facecolor(GRIS_1)
    ax.set_axisbelow(True)

    # Boxplot
    sns.boxplot(
        data=df,
        x='sexo_biologico',
        y='edad',
        palette=[AZUL_1, AQUA_1],
        ax=ax
    )

    # Título y etiquetas
    ax.set_title('Distribución de la Edad por Sexo Biológico de los clientes', fontsize=18, color=GRIS_5)
    ax.set_xlabel('', fontsize=0)  # Suprimir etiqueta del eje x
    ax.set_ylabel('Edad', fontsize=12, color=AZUL_1)
    ax.tick_params(axis='x', labelsize=12, colors=AZUL_1)
    ax.tick_params(axis='y', labelsize=12, colors=AZUL_1)

    # Eliminar bordes
    for spine in ['top', 'bottom', 'left', 'right']:
        ax.spines[spine].set_visible(False)

    # Cálculos para anotaciones
    stats = df.groupby("sexo_biologico")["edad"].agg([
        ("Mínimo", "min"),
        ("1º quartil", lambda x: x.quantile(0.25)),
        ("2º quartil", "median"),
        ("3º quartil", lambda x: x.quantile(0.75)),
        ("Máximo", "max")
    ])

    # Anotación sexo_biologico[1]
    texto1 = (
        f"Mínimo = {int(stats.iloc[1, 0])}\n"
        f"1º quartil = {int(stats.iloc[1, 1])}\n"
        f"2º quartil = {int(stats.iloc[1, 2])}\n"
        f"3º quartil = {int(stats.iloc[1, 3])}\n"
        f"Máximo = {int(stats.iloc[1, 4])}"
    )

    ax.annotate(
        texto1,
        xy=(-0.1, 45),  # punto objetivo
        xytext=(-0.4, 50),  # posición de la caja
        fontsize=12,
        color=AZUL_1,
        arrowprops=dict(facecolor=AZUL_1, arrowstyle='->'),
        bbox=dict(boxstyle="round,pad=0.4", fc=BLANCO, ec=GRIS_3, lw=1.5)
    )

    # Anotación sexo_biologico[0]
    texto2 = (
        f"Mínimo = {int(stats.iloc[0, 0])}\n"
        f"1º quartil = {int(stats.iloc[0, 1])}\n"
        f"2º quartil = {int(stats.iloc[0, 2])}\n"
        f"3º quartil = {int(stats.iloc[0, 3])}\n"
        f"Máximo = {int(stats.iloc[0, 4])}"
    )

    ax.annotate(
        texto2,
        xy=(0.9, 45),
        xytext=(0.6, 50),
        fontsize=12,
        color=AZUL_1,
        arrowprops=dict(facecolor=AZUL_1, arrowstyle='->'),
        bbox=dict(boxstyle="round,pad=0.4", fc=BLANCO, ec=GRIS_3, lw=1.5)
    )

    plt.tight_layout()
    return plt
````
