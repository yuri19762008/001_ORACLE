## **Leyendo archivos Excel**

### importando archivos Excel

```python
import pandas as pd

# Leer la hoja de cálculo de un archivo Excel
datos_co2 = pd.read_excel("emisiones_CO2.xlsx")
```

```python
datos_co2.head()  # Muestra las primeras filas del DataFrame
```

#### averiguar hojas del excel

```python
pd.ExcelFile("emisiones_CO2.xlsx").sheet_names  
# Muestra los nombres de las hojas en el archivo Excel
```

```python
percapita = pd.read_excel("emisiones_CO2.xlsx",sheet_name= 'emisiones_percapita')
```

```python
percapita.tail()  
# Muestra las ultimos filas del DataFrame
```

```python
percapita.sample(5)  
# Muestra 5 filas aleatorias del DataFrame
```

```python
fuentes = pd.read_excel("emisiones_CO2.xlsx",sheet_name= 'fuentes')
```

```python
fuentes.tail(5)
```

#### selecionar intervalo de columnas

```python
intervalo = pd.read_excel("emisiones_CO2.xlsx", sheet_name= 'emisiones_C02' , usecols= 'A:D')
intervalo
```

#### seleccionar numeros de filas 

```python
intervalo_2 = pd.read_excel("emisiones_CO2.xlsx", sheet_name= 'emisiones_C02' , usecols= 'A:D', nrows= 10)
intervalo_2
```

#### escribir archivo excel 

```python
intervalo.to_excel("emisiones_CO2_intervalo.xlsx", index= False)
```

#### leer archivo creado

```python
pd.read_excel("emisiones_CO2_intervalo.xlsx")
```
