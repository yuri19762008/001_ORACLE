#### BUSQUEDAS MAS COMPLEJAS CON REGEX

```python
# la expresion [^a-zA-Z0-9\-\'] busca cualquier caracter que no sea una letra, un digito, un guion bajo o un guion
datos['descripcion_local'] = datos['descripcion_local'].str.replace('[^a-zA-Z0-9\-\']',' ', regex=True)
```

```python
# esta expresion '(?<!\w)-(?!\w)' busca guiones que no esten precedidos ni seguidos por una letra o un digito
datos['descripcion_local'] = datos['descripcion_local'].str.replace('(?<!\w)-(?!\w)',' ', regex=True)
```
